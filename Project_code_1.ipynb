```python
import requests # import requests library to make HTTP requests
import pandas as pd # import pandas library for data manipulation and analysis
from datetime import datetime # import datetime library to work with date and time
from datetime import time # import time library to work with time
```

```python
# Initialize an empty list to store all fetched earthquake records 
all_records = []
start_year = datetime.now().year - 5 # Initialize the start year to 5 years ago from the current year
end_year = datetime.now().year # Initialize the end year to the current year
```

```python
# Loop through each month in the specified year range to fetch earthquake data and store it in a list
for year in range(start_year, end_year + 1):# Loop through each year in the specified range
    for month in range(1, 13):

        start_date = f"{year}-{month:02d}-01"

        if month == 12: # If the current month is December, set the end date to January 1st of the next year
            end_date = f"{year+1}-01-01" # Set the end date to January 1st of the next year
        else:
            end_date = f"{year}-{month+1:02d}-01" # Set the end date to the first day of the next month

        print(start_date, "→", end_date) 
```

```python
#      Define parameters for the API request (updated to fetch data for the last 12 months)
params = {
    "format": "geojson",
    "starttime": "2025-08-01",
    "endtime": "2026-08-01",
    "minmagnitude": 3 # Minimum magnitude of earthquakes to fetch
}

print(params)
```

```python
url = "https://earthquake.usgs.gov/fdsnws/event/1/query"
```

```python
# Make the API request to fetch earthquake data based on the defined parameters 
response = requests.get(url, params=params) # Make a GET request to the USGS Earthquake API with the specified parameters

if response.status_code != 200: # Check if the response status code is not 200 (indicating a failed request)
    print("Request failed:", response.text[:200])
else:
    print("Success!")
```

```python
# Explore the JSON structure 
data = response.json() # Parse the response JSON data into a Python dictionary
list(data.keys()) # show top level keys 
```

```python

# Normalize JSON data into a flat table structure using pandas 
df = pd.json_normalize(data["features"]) # Normalize the "features" key from the JSON data into a pandas DataFrame

print(df.columns)
print(df.head())
```

```python
# Clean column names

df.rename(columns=lambda x: x.replace("properties.", ""), inplace=True) 
df.rename(columns=lambda x: x.replace("geometry.", ""), inplace=True)
print(df.columns)
```

```python
# Convert time from milliseconds to datetime format 
df["time"] = pd.to_datetime(df["time"], unit="ms")
df["updated"] = pd.to_datetime(df["updated"], unit="ms")
print(df.columns)
print(df.head())
```

```python
# Split coordinates into separate columns for longitude, latitude, depth and clean column names
df[["Longitude", "Latitude", "Depth"]] = pd.DataFrame(
    df["coordinates"].tolist(),
    index=df.index
)

print(df.columns)
print(df.head())
```

```python
df["ids"] = df["ids"].str.strip(",")
df["sources"] = df["sources"].str.strip(",")
df["types"] = df["types"].str.strip(",")
```

```python
df.to_csv("earthquake_full_data_5.csv", index=False)
```
